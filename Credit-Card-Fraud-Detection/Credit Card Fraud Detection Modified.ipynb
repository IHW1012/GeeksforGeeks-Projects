```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib import gridspec
```

```python
data = pd.read_csv("creditcard.csv")
data.head()
```

```python
corrmat = data.corr()
fig = plt.figure(figsize = (12, 9))
sns.heatmap(corrmat, vmax = .8, square = True)
plt.show()
```

```python
X = data.drop(['Class'], axis = 1)
Y = data["Class"]
print(X.shape)
print(Y.shape)

xData = X.values
yData = Y.values

from sklearn.model_selection import train_test_split
xTrain, xTest, yTrain, yTest = train_test_split(
        xData, yData, test_size = 0.2, random_state = 42)
```

```python
from sklearn.ensemble import RandomForestClassifier

rfc = RandomForestClassifier()
rfc.fit(xTrain, yTrain)

yPred = rfc.predict(xTest)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, matthews_corrcoef, confusion_matrix 
accuracy = accuracy_score(yTest, yPred)
precision = precision_score(yTest, yPred)
recall = recall_score(yTest, yPred)
f1 = f1_score(yTest, yPred)
mcc = matthews_corrcoef(yTest, yPred)

print("Model Evaluation Metrics:")
print(f"Accuracy: {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")
print(f"F1-Score: {f1:.4f}")
print(f"Matthews Correlation Coefficient: {mcc:.4f}")

conf_matrix = confusion_matrix(yTest, yPred)
plt.figure(figsize=(8, 6))
sns.heatmap(conf_matrix, annot=True, fmt="d", cmap="Blues",
            xticklabels=['Normal', 'Fraud'], yticklabels=['Normal', 'Fraud'])
plt.title("Confusion Matrix")
plt.xlabel("Predicted Class")
plt.ylabel("True Class")
plt.show()
```

# Modification

```python
from sklearn.metrics import precision_recall_curve
import numpy as np

#MEmastikan hasilnya tidak berubah ketika di-run ulang
rfc_mod = RandomForestClassifier(random_state=42)
rfc_mod.fit(xTrain, yTrain)

# Ekstrak probabilitas
yPred_mod_prob = rfc_mod.predict_proba(xTest)[:, 1]

# Dapatkan semua kemungkinan threshold dari model
precisions, recalls, thresholds = precision_recall_curve(yTest, yPred_mod_prob)

# OPSI A: Mencari Threshold yang memaksimalkan F1-Score (Keseimbangan)
f1_scores = (2 * precisions[:-1] * recalls[:-1]) / (precisions[:-1] + recalls[:-1] + 1e-10)
optimal_idx_f1 = np.argmax(f1_scores)
optimal_threshold_f1 = thresholds[optimal_idx_f1]

# OPSI B: Mencari Threshold yang memaksimalkan F2-Score (Fokus Recall)
beta = 2
f2_scores = ((1 + beta**2) * precisions[:-1] * recalls[:-1]) / ((beta**2 * precisions[:-1]) + recalls[:-1] + 1e-10)
optimal_idx_f2 = np.argmax(f2_scores)
optimal_threshold_f2 = thresholds[optimal_idx_f2]

print(f"Threshold optimal (F1-Score): {optimal_threshold_f1:.4f}")
print(f"Threshold optimal (F2-Score - Prioritas Recall): {optimal_threshold_f2:.4f}")

# Kamu tinggal menggunakan angka ini untuk mapping prediksi baru
```

```python
# 1. Pilih threshold mana yang mau dipakai (misal kita ambil yang F1-Score)
final_threshold1 = optimal_threshold_f1
print(f"Menerapkan threshold: {final_threshold1:.4f}")

# 2. INI PROSES MAPPING-NYA
# Membandingkan array probabilitas dengan angka threshold, 
# lalu .astype(int) akan mengubah nilai [True, False] menjadi [1, 0]
yPred_optimal1 = (yPred_mod_prob >= final_threshold1).astype(int)

# 3. Evaluasi hasilnya menggunakan threshold baru
from sklearn.metrics import recall_score, confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

recall_optimal1 = recall_score(yTest, yPred_optimal1)
print(f"Recall dengan threshold optimal: {recall_optimal1:.4f}")

# 4. Gambar Confusion Matrix-nya
conf_matrix_optimal1 = confusion_matrix(yTest, yPred_optimal1)
plt.figure(figsize=(8, 6))
sns.heatmap(conf_matrix_optimal1, annot=True, fmt="d", cmap="Purples",
            xticklabels=['Normal', 'Fraud'], yticklabels=['Normal', 'Fraud'])
plt.title(f"Confusion Matrix (Optimal Threshold = {final_threshold1:.4f})")
plt.xlabel("Predicted Class")
plt.ylabel("True Class")
plt.show()
```

```python
# 1. Pilih threshold mana yang mau dipakai (misal kita ambil yang F2-Score)
final_threshold2 = optimal_threshold_f2
print(f"Menerapkan threshold: {final_threshold2:.4f}")

# 2. INI PROSES MAPPING-NYA
# Membandingkan array probabilitas dengan angka threshold, 
# lalu .astype(int) akan mengubah nilai [True, False] menjadi [1, 0]
yPred_optimal2 = (yPred_mod_prob >= final_threshold2).astype(int)

# 3. Evaluasi hasilnya menggunakan threshold baru
from sklearn.metrics import recall_score, confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

recall_optimal2 = recall_score(yTest, yPred_optimal2)
print(f"Recall dengan threshold optimal: {recall_optimal2:.4f}")

# 4. Gambar Confusion Matrix-nya
conf_matrix_optimal2 = confusion_matrix(yTest, yPred_optimal2)
plt.figure(figsize=(8, 6))
sns.heatmap(conf_matrix_optimal2, annot=True, fmt="d", cmap="Purples",
            xticklabels=['Normal', 'Fraud'], yticklabels=['Normal', 'Fraud'])
plt.title(f"Confusion Matrix (Optimal Threshold = {final_threshold2:.4f})")
plt.xlabel("Predicted Class")
plt.ylabel("True Class")
plt.show()
```

```python
from sklearn.ensemble import IsolationForest
from sklearn.metrics import recall_score, precision_score
import itertools

# Definisikan grid parameter
param_grid = {
    'n_estimators': [100, 200],
    'max_samples': ['auto', 256, 512],
    'contamination': [0.001, 0.01, 0.05]
}

# Buat list semua kombinasi parameter
keys, values = zip(*param_grid.items())
hyperparameter_combinations = [dict(zip(keys, v)) for v in itertools.product(*values)]

# Hanya gunakan data normal untuk training
xTrain_normal = xTrain[yTrain == 0]

best_recall = 0
best_params = None
best_precision = 0

print("Memulai iterasi tuning Isolation Forest...")
for params in hyperparameter_combinations:
    # Inisialisasi dan latih model
    iso = IsolationForest(**params, random_state=42)
    iso.fit(xTrain_normal)
    
    # Prediksi data uji
    preds_raw = iso.predict(xTest)
    preds = np.where(preds_raw == 1, 0, 1) # Mapping -1(anomali) -> 1(fraud)
    
    # Evaluasi (misalnya kita fokus pada Recall)
    current_recall = recall_score(yTest, preds)
    current_precision = precision_score(yTest, preds)
    d_precision = -(best_precision - current_precision)
    
    # Simpan parameter dengan recall terbaik (dan presisi yang tidak hancur total)
    if current_recall > best_recall and d_precision <= 0.1: 
        best_recall = current_recall
        best_precision = current_precision
        best_params = params

print(f"\nParameter Terbaik: {best_params}")
print(f"Recall Terbaik: {best_recall:.4f}")
```

```python
print("\n--- Evaluasi Isolation Forest (Tuned dengan Threshold Optimasi) ---")

# 1. Inisialisasi model final dengan parameter terbaik
best_iso = IsolationForest(**best_params, random_state=42)

# 2. Latih ulang model HANYA menggunakan data normal
best_iso.fit(xTrain_normal)

# 3. UBAH DI SINI: Ambil skor anomali mentah (bukan memakai .predict())
# Semakin negatif nilainya, semakin kuat diidentifikasi sebagai fraud
anomaly_scores = best_iso.decision_function(xTest)

# Trik Krusial: Inversi skor agar kompatibel dengan asumsi fungsi precision_recall_curve
# (Mengubah nilai negatif menjadi positif sehingga: angka besar = fraud)
inverted_scores = -1 * anomaly_scores

# Hitung precision dan recall untuk setiap kemungkinan kandidat threshold
precisions, recalls, thresholds = precision_recall_curve(yTest, inverted_scores)

# =====================================================================
# KALKULASI THRESHOLD OPTIMAL (EVALUASI KEDUA NILAI)
# =====================================================================

# Kandidat A: Mengoptimalkan F1-Score (Bobot Precision dan Recall Setara)
f1_scores = (2 * precisions[:-1] * recalls[:-1]) / (precisions[:-1] + recalls[:-1] + 1e-10)
optimal_idx_f1 = np.argmax(f1_scores)
thresh_optimal_f1 = -1 * thresholds[optimal_idx_f1] # Kembalikan ke skala asli Isolation Forest

# Kandidat B: Mengoptimalkan F2-Score (Bobot Recall Dua Kali Lebih Penting)
beta = 2
f2_scores = ((1 + beta**2) * precisions[:-1] * recalls[:-1]) / ((beta**2 * precisions[:-1]) + recalls[:-1] + 1e-10)
optimal_idx_f2 = np.argmax(f2_scores)
thresh_optimal_f2 = -1 * thresholds[optimal_idx_f2] # Kembalikan ke skala asli Isolation Forest

print(f"Hasil Optimasi Matematis:")
print(f"-> Threshold Optimal via F1-Score : {thresh_optimal_f1:.4f}")
print(f"-> Threshold Optimal via F2-Score : {thresh_optimal_f2:.4f}")

# =====================================================================
# 4 & 5. MAPPING ULANG DAN VISUALISASI SUBPLOT BERDAMPINGAN
# =====================================================================
threshold_targets = [
    ("Optimasi F1-Score", thresh_optimal_f1, "Blues"), 
    ("Optimasi F2-Score (Prioritas Recall)", thresh_optimal_f2, "Oranges")
]

# Membuat container figure dengan 2 kolom untuk perbandingan langsung
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

for idx, (name, thresh, color_map) in enumerate(threshold_targets):
    # Prosedur Mapping Berdasarkan Threshold Optimal masing-masing
    # Jika skor asli < threshold asli, maka ditandai sebagai Fraud (1)
    preds_mapped = np.where(anomaly_scores < thresh, 1, 0)
    
    # Hitung metrik evaluasi final untuk printout
    rec = recall_score(yTest, preds_mapped)
    prec = precision_score(yTest, preds_mapped)
    
    print(f"\nMetrik untuk {name}:")
    print(f"   Recall   : {rec:.4f}")
    print(f"   Precision: {prec:.4f}")
    
    # Hitung matriks dan visualisasikan pada sumbu subplot masing-masing
    conf_mat = confusion_matrix(yTest, preds_mapped)
    sns.heatmap(conf_mat, annot=True, fmt="d", cmap=color_map, ax=axes[idx],
                xticklabels=['Normal', 'Fraud'], yticklabels=['Normal', 'Fraud'])
    axes[idx].set_title(f"{name}\nThreshold: {thresh:.4f} | Recall: {rec:.4f}")
    axes[idx].set_xlabel("Predicted Class")
    axes[idx].set_ylabel("True Class")

plt.tight_layout()
plt.show()
```
